```python
from typing import List


def get_permutation(my_list: List[any], n: int) -> List[List[any]]:
    if n == 1:
        return my_list
    for i in range(n):
        get_permutation(my_list, n - 1)
        if n & 1:
            my_list[0], my_list[n - 1] = my_list[n - 1], my_list[0]
        else:
            my_list[i], my_list[n - 1] = my_list[n - 1], my_list[i]

def check_permutation(list1:List[any], list2:List[any]) -> List[List[any]]:
    permute = get_permutation(list1, len(list1))
    if list2 in permute:
        return True
    permute = get_permutation(list2, len(list2))
    if list1 in permute:
        return True
    return False

print(check_permutation([1, 2, 3], [3,1 ,2]))
```
