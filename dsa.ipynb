```python
from typing import List


def play_function(nums1: List[int], nums2: List[int]):
    """function that gets the median of combined 2 sorted arrays"""
    combined_list = sorted(nums1 + nums2)

    start = 0
    end = len(combined_list) - 1

    while start < end:
        start += 1
        end -= 1
    if start == end:
        return combined_list[start]
    return (combined_list[start] + combined_list[end]) / 2


if __name__ == "__main__":
    print(play_function([1, 1], []))
```

```python
from typing import List


def find_str_char_duplicate(s: str) -> bool:
    """function that finds the duplicate character in a string"""
    checked = set()

    for c in s:
        if c in checked:
            return True
        checked.add(c)
    return False


if __name__ == "__main__":
    print(find_str_char_duplicate("helo"))
```

```python
from typing import List
"""Check Permutation: Given two strings,
    write a method to decide if one is a
    permutation of the
other.
    """


def permute(my_list: List[int], l: int, r: int):
    if l == r:
        """ Check if l is equal to r """
        print(my_list)
    else:
        """ loop from l to r """
        for n in range(l, r):
            """ swap value at index l with value at index n
            NB: for each call in the stack the value of l will increment
            """
            my_list[l], my_list[n] = my_list[n], my_list[l]
            """ recurse on the list with l incremented by 1 """
            permute(my_list, l+1, r)
            """ backtrack the swap operation """
            my_list[l], my_list[n] = my_list[n], my_list[l]


my_list = [1, 2, 9]
permute(my_list, 0, len(my_list))
```
